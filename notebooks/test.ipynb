```python
from transformers import pipeline
 
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
 
result = classifier(
    "The product tastes great but it is never available in my local store.",
    candidate_labels=[
        "product quality",
        "taste",
        "availability",
        "price",
        "packaging"
    ],
    multi_label=True
)

for label,score in zip(result["labels"],result["scores"]):
    print(f"{label:20} {score:.2%}")
 
#print(result)
```
